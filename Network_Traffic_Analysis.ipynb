```python

import pandas as pd

traffic = pd.DataFrame({
    "Source_IP": [
        "192.168.1.10",
        "192.168.1.20",
        "10.0.0.5",
        "10.0.0.5",
        "10.0.0.5",
        "10.0.0.5"
    ],
    "Destination_Port": [
        443, 80, 21, 22, 23, 3389
    ],
    "Protocol": [
        "TCP", "TCP", "TCP",
        "TCP", "TCP", "TCP"
    ]
})

print(traffic)
```

```python

# Network intrusion detection
port_activity = traffic.groupby("Source_IP")[
    "Destination_Port"
].nunique()

print("NETWORK SECURITY ANALYSIS")
print("-------------------------")

for ip, port_count in port_activity.items():
    if port_count >= 4:
        print(f"ALERT: {ip}")
        print(f"Unique ports accessed: {port_count}")
        print("Possible port scanning activity")
    else:
        print(f"NORMAL: {ip} - {port_count} ports")
```
